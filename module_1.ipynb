```python
import json

with open("overclocking.json") as file:
    networks = json.load(file)
 
print(type(networks))
print(len(networks))    
print(networks[0])
```

```python
network = networks[0]

print("Number of users:", len(network))
print("First 5 users:", list(network.keys())[:5])
```

```python
for user, connections in network.items():
    print(user, len(connections))
```

```python
important = sorted(
    network.items(),
    key=lambda x: len(x[1]),
    reverse=True
)

print(important[:10])
```
